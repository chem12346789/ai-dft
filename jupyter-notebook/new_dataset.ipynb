```python
from ase.collections import g2

import json
import numpy as np

from pyscf.geomopt.geometric_solver import optimize
from pyscf import gto, scf, dft

dict_ = {"molecular": []}
name = g2.names

for i_name in name:
    positions = np.array(g2[i_name].get_positions())
    numbers = g2[i_name].numbers

    molecular = []
    total_ele = 0
    for i_number, atom_number in enumerate(numbers):
        atom = []
        total_ele += atom_number
        atom.append(int(atom_number))
        atom.append(float(positions[i_number][0]))
        atom.append(float(positions[i_number][1]))
        atom.append(float(positions[i_number][2]))
        molecular.append(atom)

    spin=total_ele % 2
    if spin != 0:
        name_record = f"{i_name}=SPIN=1"
    else:
        name_record = f"{i_name}"
    dict_[i_name] = molecular
    dict_["molecular"].append(name_record)
```

```python
with open("../cadft/utils/g2.json", "w") as f:
    json.dump(dict_, f)
```
